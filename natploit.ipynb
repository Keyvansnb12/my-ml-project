scatter

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6]
y = [3, 5, 4, 8, 7, 10]

plt.scatter(x, y)
plt.title('scatter plot')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```
